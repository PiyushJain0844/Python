**Histogram**

A histogram shows the distribution of numerical value.

```python
sns.histplot (
    data=titanic,
    x="age"
)

plt.show()
```

**Heatmap**

Heatmaps are extremely important for correlation analysis.

```python
correlation = titanic.select_dtypes("number").corr()

print(correlation)
```

```python
sns.heatmap (
    correlation
)
plt.show()
```

```python
sns.heatmap (
    correlation,
    annot = True
)
plt.show()
```

```python
sns.heatmap (
    correlation,
    annot = True,
    cmap = "coolwarm"
)
plt.show()
```

```python
plt.figure(figsize=(10,7))

sns.heatmap (
    correlation,
    annot = True,
    cmap = "coolwarm"
)
plt.show()
```
