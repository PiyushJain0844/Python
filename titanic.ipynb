```python
data['Gender'] = data["Gender"].replace((
    "male": 1,
    "female": 0
))
```

```python
import numpy as np 

data["SurvivalStatus"] = np.where(
    data["Survival"] == 1,
    "Survived",
    "Not Survived"
)
```

```python
data['SurvivalStatus'].value_counts()
```

```python
df1 = data[["Pclass", "Name"]]
df2 = data[["Pclass", "Survival"]]

result = pd.merge(
    df1,
    df2,
    on="Pclass",
    how="inner"
)

pd.merge(df1, df2, how="inner")
pd.merge(df1, df2, how="left")
pd.merge(df1, df2, how="right")
pd.merge(df1, df2, how="outer")
```
